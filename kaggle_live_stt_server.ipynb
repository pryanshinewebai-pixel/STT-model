### Step 3: Launch Live STT Web App on NVIDIA GPU

```python
import subprocess, time, torch

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'🚀 Starting Live STT Server on device: {device.upper()}...')

# Start server process in background
server_proc = subprocess.Popen(
    ['python3', 'scripts/test_web_app.py', '--port', '8080', '--device', device],
    cwd='STT-model',
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

# Wait 5s for models to load
time.sleep(5)
print('✅ Live STT server process running on port 8080!')
```

### Step 4: Open Public Live Link (Share with Anyone!)

```python
import subprocess, re, time

print('🔗 Creating secure Public HTTPS Tunnel...')
tunnel_proc = subprocess.Popen(
    ['/tmp/cloudflared', 'tunnel', '--url', 'http://localhost:8080'],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

public_url = None
for _ in range(40):
    line = tunnel_proc.stdout.readline()
    match = re.search(r'https://[a-zA-Z0-9-]+\.trycloudflare\.com', line)
    if match:
        public_url = match.group(0)
        break
    time.sleep(0.5)

if public_url:
    print('\n' + '=' * 70)
    print('🎉 YOUR MULTILINGUAL STT APP IS LIVE ON CLOUD GPU!')
    print(f'👉 PUBLIC LINK: {public_url}')
    print('=' * 70 + '\n')
    print('📱 Open this link on your Mobile Phone, Tablet, or PC to test!')
    print('🎙️ Click microphone, speak in Gujarati, Hindi or English, and see instant output.')
    print('⏳ Keep this cell running — it will stay active for up to 12 hours!')
else:
    print('⚠️ Could not parse URL automatically. Log output:')
    print(tunnel_proc.stdout.read())

# Keep session alive
try:
    while True:
        time.sleep(60)
except KeyboardInterrupt:
    print('\n🛑 Tunnel stopped by user.')
```
